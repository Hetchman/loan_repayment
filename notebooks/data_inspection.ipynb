# Loan repayment data inspection

This notebook shows the first stage of the project: inspecting the original CSV files before creating reusable scripts. The source data is not changed here.

## 1. Imports and file locations

```python
from pathlib import Path
import pandas as pd

project_folder = Path.cwd()

# This also works when the notebook is started from the notebooks folder.
if not (project_folder / 'Datasets').exists():
    project_folder = project_folder.parent
source_folder = project_folder / 'Datasets' / 'source'
source_folder
```

## 2. Check the files provided

There are three data files: customers, loans, and repayments.

```python
list(source_folder.glob('*.csv'))
```

## 3. Load the CSV files

I load each file separately so I can understand what it represents before joining anything.

```python
customers = pd.read_csv(source_folder / 'customers.csv')
loans = pd.read_csv(source_folder / 'loans.csv')
repayments = pd.read_csv(source_folder / 'repayments.csv')
```

## 4. Inspect the structure

The number of rows, column names, sample rows, and inferred data types help identify likely keys, dates, amounts, and categorical fields.

```python
for name, data in [('customers', customers), ('loans', loans), ('repayments', repayments)]:
    print(name.upper())
    print('Rows:', data.shape[0])
    print('Columns:', data.shape[1])
    display(data.head(3))
    print(data.dtypes)
    print()
```

## 5. Check missing values and duplicate rows

This is the first data-quality check. Missing values are expected in some optional fields, but key fields need more attention.

```python
for name, data in [('customers', customers), ('loans', loans), ('repayments', repayments)]:
    print(name.upper())
    print('Missing values:')
    print(data.isna().sum())
    print('Duplicate rows:', data.duplicated().sum())
    print()
```

## 6. Check the expected IDs

The data dictionary and column names suggest `customer_id`, `loan_number`, and `repayment_id` are identifiers. I check whether they are missing or repeated.

```python
print('Duplicate customer IDs:', customers['customer_id'].duplicated().sum())
print('Duplicate loan numbers:', loans['loan_number'].duplicated().sum())
print('Duplicate repayment IDs:', repayments['repayment_id'].duplicated().sum())
```

## 7. Inspect values that may need standardising

I inspect fields with a small number of different values. This helps identify spelling, casing, and invalid-value problems before creating validation rules.

```python
print(customers['country'].value_counts(dropna=False))
print(loans['application_state'].value_counts(dropna=False))
print(loans['is_top_up'].value_counts(dropna=False))
print(repayments['repayment_status'].value_counts(dropna=False))
```

## 8. Check relationships between files

Loans should belong to a customer. Repayments should belong to a loan.

```python
loan_customer_problems = loans[~loans['customer_id'].isin(customers['customer_id'])]
repayment_loan_problems = repayments[~repayments['loan_number'].isin(loans['loan_number'])]

print('Loans without a matching customer:', len(loan_customer_problems))
print('Repayments without a matching loan:', len(repayment_loan_problems))
```

## 9. Check monetary fields

Negative loan or repayment amounts need to be investigated before loading to the reporting tables.

```python
print('Negative loan amounts:', (loans['loan_amount'] < 0).sum())
print('Negative repayment amounts:', (repayments['amount'] < 0).sum())
```

## 10. What I created from this inspection

After repeating these checks, I moved the useful checks into two simple scripts:

- `scripts/profile_data.py` creates a Markdown data-quality report.
- `scripts/validate_data.py` checks the IDs, relationships, boolean field, amounts, and invalid repayment dates.

The scripts make the inspection repeatable before the PostgreSQL loading stage.
